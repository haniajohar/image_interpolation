Objective: Apply image interpolation and geometric transformations using library functions and pixel formulas.


Q1. (a) Download any image from the internet and include its source URL. Using OpenCV or Pillow, resize it by
1.5x with nearest-neighbor

```python
import cv2

# 1. Load image
image_path = 'eye.jpg'
img = cv2.imread(image_path)

if img is None:
    raise FileNotFoundError("Image not found. Check the path.")

# 2. Resize by 1.5x using Nearest-Neighbor Interpolation
# Method A: Using scaling factors fx and fy
resized_nn = cv2.resize(img, (0, 0), fx=1.5, fy=1.5, interpolation=cv2.INTER_NEAREST)

# Print dimensions to verify
h_orig, w_orig = img.shape[:2]
h_new, w_new = resized_nn.shape[:2]
print(f"Original Dimensions: {w_orig}x{h_orig}")
print(f"Resized Dimensions (1.5x): {w_new}x{h_new}")

# 3. Save and display
cv2.imwrite('eye_resized_nn.jpg', resized_nn)
cv2.imshow('Nearest Neighbor 1.5x', resized_nn)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
